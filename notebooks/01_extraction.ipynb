Load the raw Beijing PM2.5 dataset, inspect its structure, and confirm it is suitable for the project.

**Dataset:** PRSA_data_2010.1.1-2014.12.31.csv  
**Source:** UCI Machine Learning Repository  
**Granularity:** One row per hour, single monitoring station (US Embassy, Beijing)

```python
from pathlib import Path

import pandas as pd

PROJECT_ROOT = Path.cwd().resolve().parent if Path.cwd().resolve().name == 'notebooks' else Path.cwd().resolve()
```

## 1. Load Raw Dataset

```python
RAW_PATH = PROJECT_ROOT / 'data/raw/PRSA_data_2010.1.1-2014.12.31.csv'
df = pd.read_csv(RAW_PATH)
df.head()
```

## 2. Shape and Schema

```python
print('Rows:   ', len(df))
print('Columns:', len(df.columns))
print()
df.info()
```

## 3. Missing Values

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
pd.DataFrame({'missing_count': missing, 'missing_pct': missing_pct})
```

## 4. Basic Statistics

```python
df.describe(include='all')
```

## 5. Categorical Column Inspection

```python
# Wind direction is the only categorical column
print('cbwd value counts:')
print(df['cbwd'].value_counts())
print()
print('Note: "cv" = calm/variable wind, not a compass direction.')
```

## 6. Date Range Confirmation

```python
print('Year range: ', df['year'].min(), '—', df['year'].max())
print('Month range:', df['month'].min(), '—', df['month'].max())
print('Hour range: ', df['hour'].min(), '—', df['hour'].max())
```

## 7. Extraction Notes

- Dataset has **43,824 rows** and **13 columns** — meets the >5,000 row and >8 column requirements.
- Only `pm2.5` has missing values: **2,067 nulls (~4.7%)**. These will be handled in notebook 02.
- `cbwd` has 4 unique values: `NW`, `NE`, `SE`, `cv`. The `cv` value means calm/variable wind and is a valid category.
- The `No` column is a row index from the source file and carries no analytical value — will be dropped in cleaning.
- All other columns are complete with no nulls.
